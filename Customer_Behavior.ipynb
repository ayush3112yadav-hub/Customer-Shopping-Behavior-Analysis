```python
import pandas as pd

df = pd.read_csv("customer_behavior.csv")
print(df.head())
```

```python
# 2. View dimensions
print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
```

```python
# 3. View column names and exact data types
print("--- Column Types & Non-Null Counts ---")
df.info()
```

```python
# 4. Check for missing values per column
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
# 5. Convert all column names to lowercase for consistency
df.columns = df.columns.str.lower()
```

```python
# 6. Replace spaces in column names with underscores (_)
df.columns = df.columns.str.replace(' ', '_')
```

```python
# 7. Check the exact column names in the DataFrame
print(df.columns.tolist())
```

```python
# 8. Fill the 37 missing ratings with the median rating per Category
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# 9. Rechecking for missing values per column
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
# 10. Check for duplicate rows
print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

```python
# 11. Rename a specific column to remove unwanted characters or formatting
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
# 12. Display the updated list of column names
df.columns
```

```python
# 13. Drop exact duplicates if present
df = df.drop_duplicates()
```

```python
# 14. Adding Age group
bins = [0, 25, 40, 55, 100]
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True).astype(str)
```

```python
# 15. Purchase frequency in days
frequency_map = {
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_map)
```

```python
# 16. Quick check
print(df[['age', 'age_group', 'frequency_of_purchases', 'purchase_frequency_days']].head(10))
print(df['age_group'].value_counts())
print(df['purchase_frequency_days'].isnull().sum())   # should be 0
```

```python
# 17. Are discount_applied and promo_code_used identical?
print((df['discount_applied'] == df['promo_code_used']).all())
```

```python
# 18. Cross-tab for a closer look
print(pd.crosstab(df['discount_applied'], df['promo_code_used']))
```

```python
# 19. Drop the 'promo_code_used' column
df = df.drop(columns=['promo_code_used'])
```

```python
# 20. Show all column names that appear more than once in the DataFrame
print(df.columns[df.columns.duplicated(keep=False)])
```

```python
# 21. Remove duplicate columns, keeping only the first occurrence of each name
df = df.loc[:, ~df.columns.duplicated()]
```

```python
# 22. Verify that all column names are now unique
print(df.columns.is_unique)   # should print True
```
